# 04 - Stage 6.4: MC Dropout and split conformal

Calibrates on the held-out calibration group and evaluates on val. Reports oracle and predicted-mask results separately, including sub-2 cm metrics.

```python
import json
import subprocess
import sys
from pathlib import Path
from IPython.display import Image, display

root = Path.cwd()
final = root / "results" / "final"
for name in ("oracle", "real"):
    run_dir = final / name
    predictions = run_dir / "cls_predictions.csv"
    summary_path = run_dir / "stage64_summary.json"
    if not summary_path.exists():
        subprocess.run([
            sys.executable, "-m", "src.run_stage64", "--predictions", str(predictions),
            "--out", str(run_dir),
        ], check=True, cwd=root)
    summary = json.loads(summary_path.read_text(encoding="utf-8"))
    print(name, json.dumps({
        "auroc": summary["auroc"], "ece": summary["ece"],
        "conformal": summary["conformal"], "by_size_bucket": summary["by_size_bucket"],
    }, indent=2))
    display(Image(filename=str(run_dir / "fig_stage64_reliability.png")))
    display(Image(filename=str(run_dir / "fig_stage64_conformal.png")))
```
