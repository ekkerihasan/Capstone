# 05 - Stage 6.5: Grad-CAM++ and F-Fidelity

Explains the held-out val patches using the stage 6.3 classifier, tests salient-mask output drop against random masks, and saves the real-data figures and pass-rate summary. CUDA is selected when available.

```python
import json
import subprocess
import sys
import torch
from pathlib import Path
from IPython.display import Image, display

root = Path.cwd()
out = root / "results" / "final" / "stage65"
if not (out / "stage65_summary.json").exists():
    subprocess.run([
        sys.executable, "-m", "src.run_stage65",
        "--split", str(root / "results" / "review1_local" / "full281" / "split_281cases.json"),
        "--ckpt", str(root / "checkpoints" / "stage63" / "best.pt"),
        "--pred-masks", str(root / "results" / "final" / "masks"),
        "--group", "val", "--out", str(out),
        "--device", "cuda" if torch.cuda.is_available() else "cpu",
    ], check=True, cwd=root)

summary = json.loads((out / "stage65_summary.json").read_text(encoding="utf-8"))
print(json.dumps(summary, indent=2))
for figure in sorted(out.glob("fig_saliency_*.png")):
    display(Image(filename=str(figure)))
```
